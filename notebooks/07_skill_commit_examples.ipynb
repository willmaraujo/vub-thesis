# Some commit examples

I picked these from the 10 skills in the sample and looked at what changed in each `SKILL.md`. I wanted to check if the commit title actually describes the skill edit.

## New feature

- [Session continuity](https://github.com/Wilder1222/superomni/commit/60ddcae41ba4a57c9fe124d3ea64ec95e9e88e87): the agent gets instructions for picking up work when the user returns after a completed session.
- [Validation workflow](https://github.com/duc01226/EasyPlatform/commit/5ca0c50ee50cab2ef2750b122322e65dab120d02): `refine/SKILL.md` gets an INVEST check and more questions to ask during validation.
- [Gabb-first guidance](https://github.com/gabb-software/gabb-cli/commit/3bc5d0397703d13dd5b78095e8f7cb0881a565e0): the skill now helps the agent decide which gabb tool to use before reading code.
- [Auto-memory integration](https://github.com/autonomys/openclaw-skills/commit/d1bc4a62624242d77144c872f99b3d73cd959bcb): the title says `docs`, but the skill gains an instruction to anchor the new CID after saving a memory.
- [Team collaboration workflow](https://github.com/duc01226/EasyPlatform/commit/4f8f3500349dd95c98165fda8d6fcfebc28a2d77): the `refine` skill gets a workflow for turning an idea into a PBI.

In these cases, the “feature” is something new the skill tells the agent to do.

## Bug fixing

- [Broken Unicode](https://github.com/ericpzh/AC27LevelEditor/commit/51fb67e7c195260c8d5c715ca99ee647930a1cdd): garbled characters in the skill text were corrected.
- [Gabb structure guidance](https://github.com/gabb-software/gabb-cli/commit/17c4aa722b4c8010202c851ce1795e4c4f72cfee): the instructions about when to use `gabb_structure` were relaxed. I'm not sure if “bug fix” or “improvement” fits better here.
- [Comment marker](https://github.com/Chris-Graffagnino/creance/commit/208abd674b9b8e51d1a7cd5a84d4a920d16b50c0): the skill now says the marker counts only if it is on the comment's last line.
- [EnterPlanMode rule](https://github.com/Wilder1222/superomni/commit/40a93e2c2abc993f6426287076de1f6451e47fb4): a rule that was missing from the skill was added.
- [Reconciliation rule](https://github.com/Chris-Graffagnino/creance/commit/f6f2085257c75d62d09a60eabbff56226b2d2d41): the instruction was narrowed to merged or landed work.

## Refactoring

- [Documentation paths](https://github.com/Wilder1222/superomni/commit/4846ac0331797aca3023a2b95cdc6dfc8670bed0): the skill's links were changed after supporting docs moved to `docs/superomni/`.
- [Shared protocols](https://github.com/duc01226/EasyPlatform/commit/beba4ec75f069e73acc2bc9c16bb4d547f2cac1c): some links to shared files were removed because the instructions were already inside the skill.
- [Inlined references](https://github.com/duc01226/EasyPlatform/commit/a907aea850a9c029ab8668cd9a712fdd693955c4): reference material was brought into `refine/SKILL.md`. The commit also adds workflows, so this is not just reorganization.
- [Protocol summaries](https://github.com/duc01226/EasyPlatform/commit/44a13aa086b47f5ce6c1bc7d375cf9bd8b4a9436): short summaries were put next to the reference links. This also makes the instructions stronger.
- [Pipeline structure](https://github.com/Wilder1222/superomni/commit/fd533b66fab5cc4a30cdab8aa47cbbe22a411503): the skill gets an auto-advance rule. It is labelled as a refactor, but the workflow changes too.

## When the title is about something else

- [Refresh-root-scan fix](https://github.com/ericpzh/AC27LevelEditor/commit/87ff028917cb546efeaa9b97ca19812bea197e57): the title is about an application bug. The skill edit mostly updates its testing instructions.
- [Consolidate src/](https://github.com/ericpzh/AC27LevelEditor/commit/d365019ab921629d44c57ad6fd0edae8b67533c5): the title is about reorganizing code. The skill edit updates its description of that code.

These two are why I don't want to assume the title always describes the skill change. I haven't checked enough commits to say how common this is.
